```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:5], df.iloc[:,-1], test_size=0.2, random_state=42)

X_train.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer
```

```python
imp = SimpleImputer()
X_train_fever = imp.fit_transform(X_train[['fever']])
X_test_fever = imp.transform(X_test[['fever']])

X_train_fever.shape
```

```python
ohe = OneHotEncoder(sparse_output=False, drop='first',dtype='int32')
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city = ohe.transform(X_test[['gender','city']])

X_train_gender_city.shape
```

```python
ord_enc = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough = ord_enc.fit_transform(X_train[['cough']])
X_test_cough = ord_enc.transform(X_test[['cough']])

X_train_cough.shape
```

```python
transformer = ColumnTransformer(transformers = [
    ('tfn1', SimpleImputer(), ['fever']),
    ('tfn2', OneHotEncoder(sparse_output=False, drop='first',dtype='int32'), ['gender','city']),
    ('tfn3', OrdinalEncoder(categories=[['Mild','Strong']]), ['cough'])
    
], remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
transformer.transform(X_test).shape
```
