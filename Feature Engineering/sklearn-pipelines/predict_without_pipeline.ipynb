```python
import pickle
import pandas as pd
import numpy as np
```

```python
ohe_emb = pickle.load(open('/Users/shishir/100 days of ML/Feature Engineering/sklearn-pipelines/models/ohe_emb.pkl', 'rb'))
ohe_sex = pickle.load(open('/Users/shishir/100 days of ML/Feature Engineering/sklearn-pipelines/models/ohe_sex.pkl', 'rb'))
clf = pickle.load(open('/Users/shishir/100 days of ML/Feature Engineering/sklearn-pipelines/models/clf.pkl', 'rb'))
```

```python
arr = np.array([2,'male',31.0, 0, 0, 10.5, 'S'])
arr, arr.shape
# by default everything is stored as string. If we want to retain original datatype, use dtype = 'object'
# (7,) represents 1D array
# reshape(1,7) makes it 2-Dimensional [[2,'male',31.0, 0, 0, 10.5, 'S']]
```

```python
test_input = np.array([2,'male',31.0, 0, 0, 10.5, 'S'], dtype='object').reshape(1,7) # numpy.ndarray
test_input, test_input.shape # 1 row, 7 columns
```

```python
# Now we need to apply same preprocessing steps that we applied at the time of training
```

```python
test_input[:,1], type(test_input[:,1]) # numpy 1D array
```

```python
test_input[:,1].reshape(1,1) # 2D dataframe (1 row, 1 col)
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
test_input_sex
```

```python
test_input_emb = ohe_emb.transform(test_input[:,-1].reshape(1,1))
test_input_emb
```

```python
test_input_age = test_input[:,2].reshape(1,1)
test_input_age
```

```python
# maintaining order is very crucial
# remaining cols, age, sex, embarked
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]], test_input_age, test_input_sex, test_input_emb), axis=1)
test_input_transformed, test_input_transformed.shape
```

```python
clf.predict(test_input_transformed)
```
