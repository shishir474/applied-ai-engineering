## One hot encoding on nominal categorical variables

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
# creating a test encoder object to verify its behaviour with categories parameter
ohe_sex_tmp = OneHotEncoder(sparse_output=False)
ohe_sex_tmp.fit(X_train[['Sex']])

# Notice that sklearn sorts categories alphabetically by default.
print(ohe_sex_tmp.categories_)
```

```python
# No, it's not mandatory to pass categories.
# without categories: During fit(), sklearn automatically discovers the categories from the training data
# by default: sklearn sorts categories alphabetically by default.
# when we pass categories explicitly, we're explicitly telling sklearn 

# when would you specify categories:
# 1. you need consistent col ordering
    # without categories
    #     OneHotEncoder(sparse_output=False)
    # will produce 
    #     female  male
    #     1       0
    #     0       1
# because sklearn sorted them alphabetically.
    # With:
    #     categories=[['male','female']]
    # you get:
    # male  female
    # 1     0
    # 0     1

# 2. handle missing categories in training data.
# Suppose your training set accidentally contains only: male
# Without specifying categories:
#     ohe.fit(X_train)
# sklearn learns only:
# ['male']
# and creates just one encoded column.
# Later, when test data contains "female", you'll run into issues (unless using appropriate handle_unknown settings).
# By specifying:
#     categories=[['male','female']]

# you ensure both columns always exist.

# 3. Control drop = 'first'
# The category that gets dropped depends on the category ordering.
# categories=[['male','female']]
# male -> dropped
# female -> encoded column


# best practise: always specify categories: things are more clear
ohe_sex = OneHotEncoder(categories=[['male','female']], sparse_output=False)
```

```python
X_train_sex_ohe = ohe_sex.fit_transform(X_train[['Sex']])
X_test_sex_ohe = ohe_sex.transform(X_test[['Sex']])

X_train_sex_ohe.shape
```

```python
ohe_sex.categories_
```

```python
ohe_emb = OneHotEncoder(sparse_output=False)
```

```python
X_train_emb_ohe = ohe_emb.fit_transform(X_train_emb_imp)
X_test_emb_ohe = ohe_emb.transform(X_test_emb_imp)
```

```python
X_train_emb_ohe
```

## concatenate all the transformed cols together

```python
X_train_sex_ohe # 2 col
X_train_emb_ohe # 3 col
X_train_age_imp # 1 col
```

```python
X_train.head()
```

```python
X_train_remaining = X_train.drop(columns=['Sex', 'Embarked', 'Age'])
X_test_remaining = X_test.drop(columns=['Sex', 'Embarked', 'Age'])

X_train_remaining.head()
```

```python
X_train_remaining.values
```

```python
# now merge all of them
# X_train_remaining --> pandas dataframe
# To convert the df into ndarray --> X_train_remaining.values

X_train_transformed = np.concatenate((X_train_remaining.values, X_train_age_imp, X_train_sex_ohe, X_train_emb_ohe), axis=1)
X_test_transformed = np.concatenate((X_test_remaining.values, X_test_age_imp, X_test_sex_ohe, X_test_emb_ohe), axis=1)

X_train_transformed.shape, X_test_transformed.shape 
```

## model training

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf =DecisionTreeClassifier()
```

```python
# model training
clf.fit(X_train_transformed, y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

### pandas series

```python
s = pd.Series([10,20,30])
s
```

```python
# A Series is basically
# you can think of Series as 
    # Series = ndarray + Index

# Internally
s.values
# returns the underlying NumPy array
# and 
s.index
# returns the index
```

## Alignment Magic

```python
s1.values # gives me ndarray
```

```python
s1.index # gives me index
```

```python
# pandas series = ndarray + index
```
