```python
import pandas as pd
```

```python
df = pd.read_csv('../datasets/customer.csv')
df.head()
```

```python
# Encoding categorical variables

# In ML we need to encode categorical values. ML models expect numbers and cannot work with categories

# There are 3 ways to encode categorical data
# 1. One hot encoding - used for encoding nominal categorical features.
# 2. Label Encoding - used for target categorical cols
# 3. Ordinal Encoding - used for ordinal categorical features

# NOTE: if your target variable is categorical, use LabelEncoder.
# Label Encoding is only meant to encode target variables. 

# gender --> categorical yes, type nominal
# review --> categorical yes, type ordinal
# education --> categorical yes, type ordinal
# purchased --> categorical yes, type nominal
```

```python
df = df.iloc[:, 2:]
df.head()
```

```python
df['education'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']])
```

## train test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:2], df.iloc[:,-1], test_size=0.2, random_state=42)

X_train.shape, X_test.shape
```

```python
X_train.head()
```

```python
# fit on the train set
oe.fit(X_train)

X_train = oe.transform(X_train)
X_test = oe.transform(X_test)

X_train[:5]
```

```python
X_train = pd.DataFrame(X_train, columns=['review','education'])
X_test = pd.DataFrame(X_test, columns=['review','education'])
```

```python
oe.categories_
```

```python
# Label Encoder -- which is only used for encoding target categorical variables
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# fit on the training sets
le.fit(y_train)

# transform both training and test sets
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```
